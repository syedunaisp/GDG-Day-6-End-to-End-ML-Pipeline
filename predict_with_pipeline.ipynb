# 🚀 Inference — With Pipeline

Same task: predict a new student's score.  
But now... watch how short this is. 👇

```python
import pandas as pd
import joblib
```

## Step 1 — Load the Pipeline

One file. Everything is inside.

```python
pipeline = joblib.load("with_pipeline.pkl")
print("Pipeline loaded!")
```

## Step 2 — Create a New Student Sample

Raw data. No preprocessing needed.

```python
new_student = pd.DataFrame([{
    "study_hours"             : 6.0,
    "attendance_percentage"   : 80.0,
    "sleep_hours"             : 7.0,
    "previous_exam_score"     : 70,
    "internet_access"         : "Yes",
    "parental_education"      : "Graduate",
    "extracurricular_activity": "Yes",
    "part_time_job"           : "No",
    "motivation_level"        : "Medium"
}])

new_student
```

## Step 3 — Predict

No manual encoding. No scaling. No column alignment. Just predict.

```python
prediction = pipeline.predict(new_student)
print(f"Predicted Final Score: {prediction[0]:.2f}")
```
